## Array Concatenation and Splitting

### 1.Concatenation of arrays

```python
x = np.array([1, 2, 3])
y = np.array([3, 2, 1])
np.concatenate([x, y])
```

```python
z = [99, 99, 99]
print(np.concatenate([x, y, z])) # Without increasing the dimension
```

```python
np.vstack([x,y]) # Increasing the dimension
```

```python
x = np.array([1, 2, 3])
grid = np.array([[9, 8, 7],
                 [6, 5, 4]])
```

```python
np.vstack([x, grid])
```

```python
y = np.array([[99],
              [99]])
```

```python
np.hstack([grid, y])
```

### Splitting of arrays

```python
x = [1, 2, 3, 99, 99, 3, 2, 1]
x1, x2, x3 = np.split(x, [3, 6])
print(x1, x2, x3)
```

```python
grid = np.arange(16).reshape((4, 4))
grid
```

```python
upper, lower = np.vsplit(grid, [2])
print(upper)
print(lower)
```

```python
left, right = np.hsplit(grid, [2])
print(left)
print(right)
```

## Operators in Numpy  

```python
x = np.arange(4)
print("x =", x)
print("x + 5 =", x + 5)
print("x - 5 =", x - 5)
print("x * 2 =", x * 2)
print("x / 2 =", x / 2)
print("x // 2 =", x // 2)
print("-x = ", -x)
print("x ** 2 = ", x ** 2)
print("x % 2 = ", x % 2)
print(-(0.5*x + 1) ** 2)
np.add(x, 2)
```

```python
x = [1, 2, 3]
print("x =", x)
print("e^x =", np.exp(x))
print("3^x =", np.power(3, x))
```

```python
x = [1, 2, 4, 10]
print("x =", x)
print("ln(x) =", np.log(x))
print("log2(x) =", np.log2(x))
print("log10(x) =", np.log10(x))
```
